```python
#suppression des colonnes inutiles
colonnes_inutiles = [
    "description_produit",    #  à 100 % de valeurs manquantes
    "code_postal_livraison",  # a 86,2 % de valeurs manquantes
    "email_client",           # masquée , car les email et les motes de passes des client st toujours masqués lors du remplissage des formulaire pour plus de confidentialité
    "mot_de_passe_client",    # masquée
    "statut_produit",         # toujours égale à 0
]
data = df_c.drop(columns=colonnes_inutiles)
print(df.shape)
```

```python
# suppression des colonnes redondantes
paires = [
    ("profit", "benefice_commande"),
    ("total_ligne", "ventes_par_client"),
    ("produit_categorie_id", "categorie_id"),
    ("client_id", "commande_client_id"),
    ("produit_id", "ligne_produit_id"),
    ("prix_produit", "prix_unitaire"),
]

for copie, original in paires:
    if copie not in data.columns or original not in data.columns:
        print(f"{copie} / {original} : colonne absente, paire ignorée")
    elif (data[copie] == data[original]).all():
        data = data.drop(columns=copie)
        print(f"{copie} = {original} : identiques, copie supprimée")
    else:
        nb = (data[copie] != data[original]).sum()
        print(f"{copie} et {original} : {nb} lignes différentes, colonne conservée")

print(data.shape)
```

J’ai comparé ligne par ligne six paires de colonnes soupçonnées d’être redondantes d’après leurs noms et leurs statistiques. Les six paires se sont révélées strictement identiques sur les 180 519 lignes : benefice_commande et profit, ventes_par_client et total_ligne, categorie_id et produit_categorie_id, commande_client_id et client_id, ligne_produit_id et produit_id, prix_unitaire et prix_produit. Dans chaque cas, je n’ai conservé qu’une des deux colonnes, la seconde n’apportant aucune information supplémentaire. Cette redondance vient probablement de la fusion de plusieurs tables dans le fichier d’origine. Au total, le jeu de données passe de 53 à 42 colonnes, soit 11 colonnes supprimées, sans aucune perte d’information et sans modification du nombre de lignes

```python
#conversion des dates en types dates
for col in ["date_commande", "date_expedition"]:
    data[col] = pd.to_datetime(data[col], format="%m/%d/%Y %H:%M", errors="coerce")

print(data[["date_commande", "date_expedition"]].dtypes)
print(data[["date_commande", "date_expedition"]].isna().sum())
print(data["date_commande"].min(), "->", data["date_commande"].max())
```

```python
#nettoyons les texte: suppression des espaces au début et à la fin, et réduction des espaces doubles à un seul.

# Sélectionner les colonnes textuelles, sauf les dates
colonnes_texte = [
    col for col in data.select_dtypes(include=["object", "string"]).columns #repère automatiquement les colonnes de texte ; Les deux dates sont exclues, car elles ne sont pas du texte à nettoyer.
    if col not in ["date_commande", "date_expedition"]
]

print(f"{len(colonnes_texte)} colonnes textuelles à nettoyer.")
```

```python
# Nettoyer les espaces dans chaque colonne
for col in colonnes_texte:
    anciennes_valeurs = data[col].copy()

    data[col] = data[col].str.strip().str.replace(r"\s+", " ", regex=True) #.str.strip() enlève les espaces au début et à la fin , .str.replace(r"\s+", " ", regex=True) remplace plusieurs espaces à la suite par un seul.

    # Compter les valeurs qui ont été modifiées
    nb_corrigees = (
        (anciennes_valeurs != data[col]) & anciennes_valeurs.notna()
    ).sum()

    if nb_corrigees > 0:
        print(f"{col} : {nb_corrigees} valeurs corrigées")

print("Nettoyage des textes terminé.")
```

Le nettoyage des textes a porté sur les 20 colonnes textuelles : suppression des espaces en début et en fin de valeur, et réduction des espaces multiples à un seul. Au total, 25 526 valeurs ont été corrigées dans 6 colonnes, les 14 autres étant déjà propres. La colonne la plus touchée est region_livraison, avec 17 925 valeurs corrigées, soit près de 10 % des lignes ; viennent ensuite rue_client (3 036), produit (1 774), categorie (1 543), image_produit (886) et departement (362). Ces défauts sont invisibles à l’œil nu mais lourds de conséquences : pour un ordinateur, deux libellés qui ne diffèrent que par un espace sont deux catégories distinctes. Sans cette correction, les regroupements par région, par catégorie ou par produit auraient donné des totaux fragmentés, et l’analyse des causes de retard en aurait été faussée

Je n’ai pas eu recours à la médiane ni au mode, méthodes habituelles d’imputation, car elles ne s’appliquent qu’aux quantités et aux catégories. Les deux colonnes concernées sont d’une autre nature : le code postal est un identifiant géographique et le nom une donnée propre à chaque personne. Leur attribuer la valeur la plus fréquente aurait créé des informations fausses, par exemple en rattachant des clients californiens au code postal majoritaire du fichier. »

Il reste à corriger le décalage : envoyez-moi le tableau du bloc précédent, celui qui affiche rue_client et les colonnes voisines.

```python
#controle de coherence

regles = {
    "date de commande invalide": data["date_commande"].isna(), #Une commande sans date ne peut être ni située dans le temps ni mesurée
    "date d'expédition invalide": data["date_expedition"].isna(),
    "expédition antérieure à la commande": data["date_expedition"] < data["date_commande"], #On ne peut pas expédier un article avant qu’il soit commandé
    "quantité nulle ou négative": data["quantite"] <= 0,
    "vente négative": data["ventes"] < 0,
    "remise négative ou supérieure à la vente": (data["remise"] < 0) | (data["remise"] > data["ventes"]),
    "délai négatif": (data["jours_reels"] < 0) | (data["jours_prevus"] < 0),
    "risque de retard différent de 0 ou 1": ~data["risque_retard"].isin([0, 1]),
}

data["motif_rejet"] = ""
for motif, masque in regles.items():
    print(f"{motif:<45} : {masque.sum()} ligne(s)")
    data.loc[masque & (data["motif_rejet"] == ""), "motif_rejet"] = motif

rejets = data[data["motif_rejet"] != ""].copy()
data = data[data["motif_rejet"] == ""].drop(columns="motif_rejet").copy()

print()
print("Lignes rejetées  :", len(rejets))
print("Lignes conservées :", len(data))
```

J’ai soumis les 180 519 lignes à huit contrôles de cohérence, chacun correspondant à une situation impossible dans la réalité : date invalide, expédition antérieure à la commande, quantité nulle ou négative, vente négative, remise négative ou supérieure au montant de la vente, délai négatif, indicateur de retard différent de 0 ou 1. Aucun contrôle n’a relevé d’anomalie : aucune ligne n’a été rejetée et les 180 519 lignes sont conservées. Ce résultat atteste de la cohérence interne du jeu de données et de la réussite de la conversion des dates. Il ne signifie pas pour autant que toutes les valeurs sont normales : ces contrôles détectent l’impossible, non l’inhabituel. Les pertes supérieures au montant de la vente, valides sur le plan logique mais atypiques, restent à signaler. Le dispositif de rejet est conservé dans le script, afin que tout fichier futur contenant des incohérences soit traité automatiquement, avec conservation du motif de chaque rejet.

```python
#creation des colonnes calculés
# 1. Écart entre le délai réel et le délai prévu
data["ecart_jours"] = data["jours_reels"] - data["jours_prevus"]

# 2. Expédition annulée : 1 = oui, 0 = non
data["expedition_annulee"] = (data["statut_livraison"] == "Shipping canceled").astype(int)

# 3. Livré à temps (à l'heure ou en avance) : 1 = oui, 0 = non
data["livre_a_temps"] = data["statut_livraison"].isin(
    ["Shipping on time", "Advance shipping"]).astype(int)

# 4 et 5. Année et mois de la commande
data["annee_commande"] = data["date_commande"].dt.year
data["mois_commande"] = data["date_commande"].dt.strftime("%Y-%m")

# Arrondi des montants à 2 décimales (sans toucher aux coordonnées)
montants = [c for c in data.select_dtypes(include="float").columns
            if c not in ("latitude", "longitude")]
data[montants] = data[montants].round(2)

print(data.shape)
```

```python
print("Écart moyen (jours)        :", round(data["ecart_jours"].mean(), 2))
print("Expéditions annulées       :", data["expedition_annulee"].sum())
print("Lignes livrées à temps     :", data["livre_a_temps"].sum())

expediees = data[data["expedition_annulee"] == 0]
print("Taux de livraison à temps  :", round(expediees["livre_a_temps"].mean() * 100, 1), "%")
print("Période :", data["mois_commande"].min(), "->", data["mois_commande"].max())
```

Cinq colonnes ont été créées pour préparer le calcul des indicateurs. ecart_jours mesure la différence entre le délai réel et le délai prévu : elle vaut en moyenne 0,57 jour, ce qui confirme que les expéditions prennent un peu plus d’une demi-journée de plus que prévu. expedition_annulee identifie les 7 754 lignes annulées, qui seront exclues des indicateurs de délai puisqu’elles n’ont jamais été livrées. livre_a_temps repère les 73 788 lignes livrées à l’heure ou en avance : rapportées aux 172 765 lignes réellement expédiées, elles donnent un taux de livraison à temps de 42,7 %, ce qui signifie que 57,3 % des expéditions arrivent en retard. Enfin, annee_commande et mois_commande montrent que les données couvrent 37 mois, de janvier 2015 à janvier 2018, période sur laquelle l’évolution des retards pourra être suivie. Ces résultats recoupent exactement ceux du diagnostic, ce qui valide les calculs. Le jeu de données compte désormais 47 colonnes pour 180 519 lignes.
