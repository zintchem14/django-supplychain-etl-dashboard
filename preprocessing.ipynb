```python
import pandas as pd

# 1. Charger le fichier d'origine
df = pd.read_csv('etl/DataCoSupplyChainDataset.csv', encoding='latin1')

# 2. Créer explicitement une colonne d'index unique pour chaque ligne (article de commande)
df['row_unique_id'] = range(1, len(df) + 1)

# 3. Sauvegarder dans un nouveau CSV propre
df.to_csv('etl/DataCo_clean_items.csv', index=False, encoding='utf-8')
print(f"Fichier propre généré avec succès : {len(df)} lignes prêtes !")
```
